# Data cleaning

**What:** Turn raw GDB (and optional spreadsheet) files into **GeoParquet** under `data/clean/`.

**Why Parquet:** Keeps geometry in a standard way so the next step (`ingest.py`) can load shapes into DuckDB without losing them.

**How:** All logic lives in `scripts/clean_data.py`. This notebook just runs that script so you have a single place to edit code and a simple “Run all” path here.

```python
import subprocess
import sys
from pathlib import Path

# If Jupyter’s working folder is notebooks/, step up to the repo root.
ROOT = Path.cwd().resolve()
if not (ROOT / "scripts" / "clean_data.py").is_file():
    ROOT = ROOT.parent

subprocess.run([sys.executable, str(ROOT / "scripts" / "clean_data.py")], cwd=ROOT, check=True)
```
